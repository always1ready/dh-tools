## Вариант 2: Топография террора в Москве

### Контекст и источники

В основе задания лежит датасет `terror_topography.csv`, содержащий информацию о 70 местах в Москве, напрямую связанных с политическими репрессиями: тюрьмах (56 объектов), местах расстрелов (7 объектов) и местах захоронений (7 объектов). Файл доступен для скачивания по [ссылке]().

### Техническая часть (отработка навыков с семинара)

1. Загрузите CSV-файл в DataFrame. Обратите внимание, что разделителем в данном файле является точка с запятой, поэтому при загрузке необходимо указать параметр `sep=';'`. Выведите размер таблицы, типы столбцов и первые 5 строк.
2. Посчитайте, сколько объектов каждого типа (`type`) представлено в данных. Попробуйте отфильтровать данные по словам в названии (`тюрьма`, `концлагерь`, `спецобъект` и проч.), добавьте эту информацию в отдельный столбец. Постройте столбчатую диаграмму распределения типов объектов с помощью библиотеки `matplotlib`.
3. Постройте карту всех 70 объектов с помощью `folium`. Раскрасьте маркеры в зависимости от их типа: места заключения — синий цвет, места расстрелов — красный, места захоронений — чёрный.
    Дополнительные задания для получения оценки 10:
    
    3a. В всплывающее окно (`popup`) добавьте название объекта (`name`) и его адрес (`address`).
    3b. Постройте карту с плагином `MarkerCluster`, посмотрите, где и как группируются разные типы объектов.
4. Постройте тепловую карту (`HeatMap`). Определите, где преимущественно сконцентрированы объекты — в историческом центре города или на его окраинах?
